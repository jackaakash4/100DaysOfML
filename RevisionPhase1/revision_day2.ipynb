```python
import numpy as np

X = np.array([
    [10, 20, 30],
    [15, 25, 35],
    [20, 30, 40],
    [25, 35, 45],
    [30, 40, 50]
])

mean = X.mean(axis=0)
std = X.std(axis=0)

X_scaled = (X - mean) / std
print(mean, std, X_scaled, mean.shape, std.shape, X_scaled.shape)
```
