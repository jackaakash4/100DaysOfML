```python
"""
Write a simple number guessing game:
1. Pick a "secret number" (just hardcode it for now, e.g. 7 — random.randint comes later)
2. Ask the user to guess, in a loop
3. If guess is too high, print "Too high"
4. If guess is too low, print "Too low"
5. If correct, print "Correct!" and stop the loop
6. Keep track of and print how many attempts it took
"""

secret_num = 7
count = 0
isCorrect = False

while not isCorrect:
    guess = int(input("Enter a guess number"))
    count = count + 1
    if guess == secret_num:
        isCorrect = True
        print("Correct!")
    elif guess < secret_num:
        print("Too low")
    else: 
        print("Too high")
        
print("No of attempt: ", count)

        
```

```python
"""
Write a function fizzbuzz_value(n) that:
- Returns "FizzBuzz" if n is divisible by both 3 and 5
- Returns "Fizz" if divisible by 3
- Returns "Buzz" if divisible by 5
- Otherwise returns str(n)

Then write a loop from 1 to 20 that calls this function and prints each result.
"""

def fizzbuzz_value(n):
    if n % 3 == 0 and n % 5 == 0:
        return "FizzBuzz"
    elif n % 3 == 0:
        return "Fizz"
    elif n % 5 == 0:
        return "Buzz"
    else:
        return n

for i in range(1, 21, 1):
    print(fizzbuzz_value(i))
```

```python
"""
Refactor your Day 2 guessing game into functions:

1. def get_guess() -> asks user for input, returns the int
2. def check_guess(guess, secret) -> returns "correct", "high", or "low" (a string, don't print inside this function)
3. def play_game(secret) -> contains the while loop, calls the two functions above, 
   prints appropriate messages based on what check_guess returns, and returns the 
   final attempt count

Then call: attempts = play_game(7)
           print(f"You won in {attempts} attempts!")
"""
def get_guess():
    guess = int(input("Enter your guess: "))
    return guess

def check_guess(guess, secret):
    if guess == secret:
        return "Correct!"
    elif guess < secret:
        return "Too low"
    else: 
        return "Too high"


def play_game(secret):
    correctAnswer = False
    count = 0
    while not correctAnswer:
        answer = get_guess()
        check = check_guess(answer, secret)
        if check == "Correct!":
            correctAnswer = True
        print(check)
        count += 1
    return count

attempts = play_game(7)
print(f"You won in {attempts} attempts")
```

```python
"""
Build a small "student database" using a LIST OF DICTIONARIES:

students = [
    {"name": "Jack", "age": 19, "grades": [85, 90, 78]},
    {"name": "Priya", "age": 20, "grades": [92, 88, 95]},
    ... (add 2-3 more yourself)
]

Then write code that:
1. Loops through all students and prints their name and AVERAGE grade
   (hint: sum(list) / len(list))
2. Finds and prints the name of the student with the HIGHEST average grade

"""
students = [
    {"name": "Jack", "age": 19, "grades": [85, 90, 78]},
    {"name": "Priya", "age": 20, "grades": [92, 88, 95]},
    {"name": "John", "age": 24, "grades": [100, 99, 98]},
    {"name": "Alex", "age": 40, "grades": [70, 80, 90]}
]

for student in students:
    print(student["name"], sum(student["grades"])/len(students))

#name of student with highest average grade
for index, student in enumerate(students):
    maximum = 0
    i = 0
    while index < len(students):
        average = sum(student["grades"])/len(student)
        if average > maximum:
            maximum = average
            i = index
print(f"Name: {students[i]["name"]} with highest average: {maximum}")
```
