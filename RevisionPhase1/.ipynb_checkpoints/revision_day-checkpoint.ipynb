```python
#Phase one revision

#Challenge 1 : Python
#list comprehension
#data
accuracies = [0.81, 0.92, 0.87, 0.95, 0.73]

good_scores = [accuracy for accuracy in accuracies if accuracy > 0.90]
print("Good_scores i.e. accuracy more than 0.90 ", good_scores)
```

```python
#Dictionary comprehension
dict1 = {
    f"model_{i+1}": acc
    for i, acc in enumerate(accuracies)
}
print("Dictionary comprehension: ", dict1)
```

```python
#lambda + map() -> iterate through all the element of the list
percentage = list(
    map(lambda x: x * 100, accuracies)
)
print("Converted accuracy into percentage", percentage)
```

```python
#bonus challenge
status_level = [
    "Production" if acc > 0.90 else "Need Improvement"
    for acc in accuracies
]
print("Status level of model: ", status_level)
```

```python
#filter() -> it only filters the matching values
good_scores = list(
    filter(lambda x: x > 0.90, accuracies)
)
print("Filtered accuracies of models: ", good_scores)
```

```python
#zip() -> combine related datasets
score = [0.81, 0.92, 0.87, 0.95, 0.73]
model = ["A", "B", "C"]

for model, score in zip(model, score):
    print(model, score)
```

```python
#Challenge 2: Numpy
import numpy as np
X = np.array([
    [10, 100],
    [20, 200],
    [30, 300],
    [40, 400],
    [50, 500]
])
print(X)
print(type(X))
```

```python
print("Mean: ", X.mean(axis = 0))
print("SD: ", X.std(axis = 0))
print("Normalize: ", (X-X.mean(axis = 0)/X.std(axis = 0)))
```

```python
#Pandas
import pandas as pd

df = pd.DataFrame({
    "model": ["A", "B", "C", "D", "E"],
    "architecture": [
        "CNN", "Transformer",
        "CNN", "Transformer", "CNN"
    ],
    "accuracy": [0.82, 0.94, 0.88, 0.96, 0.79],
    "latency": [30, 80, 35, 90, 25]
})
""" mean accuracy
median accuracy
standard deviation"""

print(df)

```

```python
print("Mean accuracy : ", df['accuracy'].mean(axis = 0))
```

```python
print("Median accuracy: ", df['accuracy'].median(axis = 0))
```

```python
print("Standard deviation: ", df['accuracy'].std(axis = 0))
```

```python
#Filter
filtered_data = df[(df['accuracy'] > 0.90) & (df['architecture'] == "Transformer")]
print("Filtered data: ", filtered_data)
```

```python
#Calculating average accuracy by architecture using groupby()

average = df.groupby('architecture')['accuracy'].mean()
print("Average by architecture: ", average)
```

```python
#Visualization
import matplotlib.pyplot as plt

plt.hist(df['accuracy'])
plt.show()
```

```python
plt.boxplot(df['latency'])
plt.show()
```

```python
#Statistic and Feature engineering
#data
import numpy as np
import pandas as pd
time = np.array([
    10, 11, 12, 10, 13,
    11, 12, 14, 100
])
times = pd.DataFrame(time, columns = ["time"])
print("Dataset information: \n", times)
print("Mean: ", times.mean(), "Median: ", times.median(), "Standard Deviation: ", times.std())
```

```python
times['log_time'] = np.log1p(times['time'])
print(times)
```

```python
times['time_squared'] = times['time'] ** 2
print(times)
```
