```python
#Phase one revision

#list comprehension
#data
accuracies = [0.81, 0.92, 0.87, 0.95, 0.73]

good_scores = [accuracy for accuracy in accuracies if accuracy > 0.90]
print("Good_scores i.e. accuracy more than 0.90 ", good_scores)
```

```python
#Dictionary comprehension
dict1 = {
    f"model_{i+1}": acc
    for i, acc in enumerate(accuracies)
}
print("Dictionary comprehension: ", dict1)
```

```python
#lambda + map()
percentage = list(
    map(lambda x: x * 100, accuracies)
)
print("Converted accuracy into percentage", percentage)
```
