```python
"""
Write a simple number guessing game:
1. Pick a "secret number" (just hardcode it for now, e.g. 7 — random.randint comes later)
2. Ask the user to guess, in a loop
3. If guess is too high, print "Too high"
4. If guess is too low, print "Too low"
5. If correct, print "Correct!" and stop the loop
6. Keep track of and print how many attempts it took
"""

secret_num = 7
count = 0
isCorrect = False

while not isCorrect:
    guess = int(input("Enter a guess number"))
    count = count + 1
    if guess == secret_num:
        isCorrect = True
        print("Correct!")
    elif guess < secret_num:
        print("Too low")
    else: 
        print("Too high")
        
print("No of attempt: ", count)

        
```

```python
"""
Write a function fizzbuzz_value(n) that:
- Returns "FizzBuzz" if n is divisible by both 3 and 5
- Returns "Fizz" if divisible by 3
- Returns "Buzz" if divisible by 5
- Otherwise returns str(n)

Then write a loop from 1 to 20 that calls this function and prints each result.
"""

def fizzbuzz_value(n):
    if n % 3 == 0 and n % 5 == 0:
        return "FizzBuzz"
    elif n % 3 == 0:
        return "Fizz"
    elif n % 5 == 0:
        return "Buzz"
    else:
        return n

for i in range(1, 21, 1):
    print(fizzbuzz_value(i))
```

```python
"""
Refactor your Day 2 guessing game into functions:

1. def get_guess() -> asks user for input, returns the int
2. def check_guess(guess, secret) -> returns "correct", "high", or "low" (a string, don't print inside this function)
3. def play_game(secret) -> contains the while loop, calls the two functions above, 
   prints appropriate messages based on what check_guess returns, and returns the 
   final attempt count

Then call: attempts = play_game(7)
           print(f"You won in {attempts} attempts!")
"""
def get_guess():
    guess = int(input("Enter your guess: "))
    return guess

def check_guess(guess, secret):
    if guess == secret:
        return "Correct!"
    elif guess < secret:
        return "Too low"
    else: 
        return "Too high"


def play_game(secret):
    correctAnswer = False
    count = 0
    while not correctAnswer:
        answer = get_guess()
        check = check_guess(answer, secret)
        if check == "Correct!":
            correctAnswer = True
        print(check)
        count += 1
    return count

attempts = play_game(7)
print(f"You won in {attempts} attempts")
```

```python
"""
Grade report generator. Given:

names  = ["Jack", "Priya", "John", "Alex"]
scores = [72, 91, 58, 85]

Write code that:
1. Uses zip to pair each name with a score and builds a dictionary {name: score}
2. Uses a dict comprehension to build a new dictionary of only the students
   who passed (score >= 60)
3. Uses enumerate (starting at 1) to print a numbered list of the passing
   students, sorted from highest to lowest score, formatted like:
   1. Priya - 91
   2. Alex - 85
   3. Jack - 72
4. Wrap steps 2 and 3 in a function that takes the score dictionary and a
   passing mark, and RETURNS the list of passing names in ranked order
   (do the printing outside the function)
   """
names  = ["Jack", "Priya", "John", "Alex"]
scores = [72, 91, 58, 85]
dict1 = {}

for name, score in zip(names, scores):
    dict1[name] = score
print(dict1)


dict_pass = {name: dict1[name] for name in dict1 if dict1[name] >= 60 }
print(dict_pass)

sorted_dict_pass = sorted(dict_pass, key = lambda x: dict_pass[x], reverse = True)
print(sorted_dict_pass)


for index, name in enumerate((sorted_dict_pass), start = 1):
    print(f"{index}. {name} - {dict_pass[name]}")


#function
"""
4. Wrap steps 2 and 3 in a function that takes the score dictionary and a
   passing mark, and RETURNS the list of passing names in ranked order
   (do the printing outside the function)
"""

def get_ranked_passing(scores_dict, passing_mark):
    sorted_dict = sorted(scores_dict, key = lambda x: scores_dict[x], reverse = True)
    ranked_names = [name for name in sorted_dict if scores_dict[name] >= passing_mark]
    return ranked_names

ranked = get_ranked_passing(dict1, 60)
for index, name in enumerate(ranked, start = 1):
    print(f"{index}. {name} - {dict1[name]}")


```
