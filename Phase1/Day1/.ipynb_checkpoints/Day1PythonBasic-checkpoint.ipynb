```python
#Day1
#Concept of object reference and shallow copy
x = [10, 20]
y = x
z = x.copy()

x.append(30)
y.append(40)
z.append(50)

print(x)
print(y)
print(z)
```

```python
#Functions and Mutability
print(["AI"] + ["ML"])
```

```python
def change(x):
    x.append(4)

a = [1, 2, 3]
change(a)

print(a)
```

```python
def change(x):
    x = x + [4]

a = [1, 2, 3]
change(a)

print(a)
```

```python
def process_data(data):
    data = data.copy()
    data.append(100)
    return data

numbers = [1, 2, 3]

result = process_data(numbers)

print(numbers)
print(result)
```

```python
def add_item(item, items=[]):
    items.append(item)
    return items

a = add_item("Python")
b = add_item("ML")
c = add_item("AI")

print(a)
print(b)
print(c)
```

```python
#Challenge
"""
Mini Challenge

Let's see if you can apply everything we've covered.

Write a function:

def clean_numbers(numbers):
    ...

Requirements:

It must not modify the original list.
Create a copy of the list.
Remove all negative numbers.
Return the cleaned list.
Do not use NumPy or Pandas yet.

Example:

Input:  [10, -5, 20, -3, 30]

Output: [10, 20, 30]
Your task

Write the function yourself.
"""

def clean_numbers(numbers):
    number = numbers.copy()
    clean_number = []
    for n in number:
        if n >= 0:
            clean_number.append(n)

    return clean_number
    


data =  [10, -5, 20, -3, 30]
print(clean_numbers(data))
```
