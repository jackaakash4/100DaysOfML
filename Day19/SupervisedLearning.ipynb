```python
#Supervised learning
print("Supervised Learning: Linear Regression")
```

```python
#Step1: Create Synthetic dataset
from sklearn.datasets import make_regression
import numpy as np

X, y = make_regression(
    n_samples = 1000,
    n_features = 50,
    n_informative = 10,
    noise = 20,
    random_state = 42
)
```

```python
#Step 2: Add multicollinearity
X = np.column_stack(
    [
        X,
        X[:, 0] + np.random.normal(0, 0.1, 1000),
        X[:, 1] + np.random.normal(0, 0.1, 1000),
        X[:, 2] + np.random.normal(0, 0.1, 1000)
    ]
)
```

```python
#Step3: Train-test split
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = (
    train_test_split(
        X,
        y,
        test_size = 0.2,               #80-20 split
        random_state = 42
    )
)
```

```python
from sklearn.linear_model import LinearRegression

lr = LinearRegression()
lr.fit(X_train, y_train)
```

```python
#train ridge
from sklearn.linear_model import Ridge

ridge = Ridge(alpha = 1.0)
ridge.fit(X_train, y_train)
```

```python
#Train lasso
from sklearn.linear_model import Lasso

lasso = Lasso(alpha = 0.1)
lasso.fit(X_train, y_train)
```

```python
#compare r square scores
print("Linear:", lr.score(X_test, y_test))

print("Ridge:", ridge.score(X_test, y_test))

print("Lasso:", lasso.score(X_test, y_test))
```

```python
#inspect lasso feature selection
print(lasso.coef_)
```

```python
#Count removed features
np.sum(lasso.coef_ == 0)
#This is one of the coolest things about lasso
```

```python
#visualize coefficient
import matplotlib.pyplot as plt

plt.figure(figsize = (12, 5))
plt.plot(lr.coef_, label = "Linear")
plt.plot(ridge.coef_, label = "Ridge")
plt.plot(lasso.coef_, label = "Lasso")
plt.legend()
plt.show()
```
