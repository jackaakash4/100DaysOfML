```python
import requests

response = requests.get("https://jsonplaceholder.typicode.com/posts/1")

print(response.status_code)
data = response.json()
print(data)
```

```python
import requests

response = requests.get("https://jsonplaceholder.typicode.com/posts/1", params = {"limit": 10}, timeout = 10)

print(response.status_code)
data = response.json()
print(data)
```

```python
import requests

try:
    response = requests.get("https://jsonplaceholder.com/posts/1", timeout = 10)
    response.raise_for_status()
    data = response.json()
except requests.exceptions.RequestException as e:
    print(f"Request failed {e}")
    data = None
print(data)

```

```python
"""
1. Write fetch_users() that does a GET to that URL with a timeout, calls
   raise_for_status(), and RETURNS the parsed list (return None on failure,
   using try/except RequestException).
2. Print how many users came back, and the name and email of the first one.
3. Predict first: what status code do you expect for
   https://jsonplaceholder.typicode.com/users/9999 (a user that doesn't exist)?
   Then try it and see.
   """

def fetch_user(url):
    try:
        response = requests.get(url, timeout = 10)
        response.raise_for_status()
        data = response.json()
    except requests.exceptions.RequestException as e:
        print(f"Request failed: {e}")
        data = None
    return data


datas = fetch_user("https://jsonplaceholder.typicode.com/users")
print(datas)
    
        
```

```python
#Debugging challenge
import requests
import os


API_KEY = os.environ.get("sk-abc123")


def get_scores(student_id):
    try:
        response = requests.get(
        f"https://api.example.com/students/{student_id}/scores",
        headers={"Authorization": f"Bearer {API_KEY}"}, timeout = 10)
        response.raise_for_status()
        body = response.json()
        
    except requests.excections.RequestException as e:
        print(f"Request error: {e}")
        return None
    
    if "scores" not in body:
        raise ValueError(f"Unexpected response for student {student_id}: {body}")
    return body["scores"]
```
