```python
import requests

response = requests.get("https://jsonplaceholder.typicode.com/posts/1")

print(response.status_code)
data = response.json()
print(data)
```

```python
import requests

response = requests.get("https://jsonplaceholder.typicode.com/posts/1", params = {"limit": 10}, timeout = 10)

print(response.status_code)
data = response.json()
print(data)
```

```python
import requests

try:
    response = requests.get("https://jsonplaceholder.com/posts/1", timeout = 10)
    response.raise_for_status()
    data = response.json()
except requests.exceptions.RequestException as e:
    print(f"Request failed {e}")
    data = None
print(data)

```
