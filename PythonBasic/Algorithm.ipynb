```python
"""
1. Write linear_search(items, target) from scratch (don't look at mine above — 
   write it yourself) — should return the index of target, or -1 if not found



   """

def linear_search(items, target):
    for key, i in enumerate(items):
        if i == target:
            return key
    return -1

sorted_list = [2, 5, 8, 12, 16, 23, 38, 45, 56, 72, 91]
print(linear_search(sorted_list, 23))
print(linear_search(sorted_list, 100))
```

```python
"""
2. Write binary_search(sorted_items, target) yourself as well
"""
def binary_search(sorted_items, target):
    low = 0
    high = len(sorted_items) - 1

    while low <= high:
        mid = (low + high)//2
        if sorted_items[mid] == target:
            return mid
        elif sorted_items[mid] < target:
            low = mid + 1
        else:
            high = mid - 1
    return -1

print(binary_search(sorted_list, 23))
print(binary_search(sorted_list, 100))

    
```

```python
""" Build challenge - Basic sorting """

def bubble_sort(items):
    sorted_list = []
    n = len(items)
    for i in range(0, n-1, 1):
        minimum = items[i]
        for j in range (i+1, n-2, 1):
            if items[j] < items[i]:
                minimum = items[j]
        sorted_list.append(minimum)
    return sorted_list
numbers = [5, 2, 8, 1, 9, 3]
#bubble_sort(numbers)

print("Sorted array: ", bubble_sort(numbers))
                
            
```
