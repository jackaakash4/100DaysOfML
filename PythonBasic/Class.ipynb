```python
"""
Write a Student class with:
1. __init__(self, name, age, grades) — stores all three as attributes
2. A method average(self) — returns the average grade (you've written 
   this exact calculation many times now)
3. A method highest(self) — returns the highest single grade
4. A method summary(self) — returns a formatted string like:
   "Jack (19): avg 84.33, highest 90"

Then create 3-4 Student objects and print each one's summary().
"""

class Student:
    def __init__(self, name, age, grades):
        self.name = name
        self.age = age
        self.grades = grades
    def average(self):
        return sum(self.grades) / len(self.grades)

    def highest(self):
        return max(self.grades)
        
    def summary(self):
        return f"{self.name} ({self.age}): avg {self.average():.2f}, highest {self.highest()}" 

john = Student("John", 20, [40, 60, 80])
print(john.summary())

jack = Student("Jack", 30, [80, 90, 100])
print(jack.summary())

feifei = Student("FeiFei", 45, [90, 90, 90])
print(feifei.summary())
```

```python
"""
Create a class GraduateStudent that inherits from your Student class:
1. It should have everything Student has (name, age, grades)
2. Plus a new attribute: thesis_title
3. Override summary() so it also includes the thesis title, but REUSE 
   Student's summary() via super() rather than rewriting the whole string
   """
class Student:
    def __init__(self, name, age, grades):
        self.name = name
        self.age = age
        self.grades = grades

    def summary(self):
        return f"Your name is {self.name}, age is {self.age} and your grades are {self.grades}"

class GraduateStudent(Student):
    def __init__(self, name, age, grades, thesis_title):
        super().__init__(name, age, grades)
        self.thesis_title = thesis_title
    def summary(self):
        base = super().summary()
        return f"{base} - Thesis title: {self.thesis_title}"

jack = GraduateStudent("Jack", 20, [80, 90, 100], "AI in education")
print(jack.summary())
        
```
