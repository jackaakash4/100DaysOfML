```python
#2
import pandas as pd
messy_data = pd.DataFrame({
    "name": ["Jack", "Priya", "John", "Jack", "Alex"],
    "age": [19, 20, None, 19, 40],
    "city": ["New York", "new york ", "LA", "New York", "LA "],
    "score": [85, 92, 78, 85, 1000]   # 1000 is clearly a data entry error
})
print("Finding duplicated values: ", messy_data.duplicated())

non_duplicate_data = messy_data.drop_duplicates()
print(non_duplicate_data)
```

```python
#4
import pandas as pd
messy_data = pd.DataFrame({
    "name": ["Jack", "Priya", "John", "Jack", "Alex"],
    "age": [19, 20, None, 19, 40],
    "city": ["New York", "new york ", "LA", "New York", "LA "],
    "score": [85, 92, 78, 85, 1000]   # 1000 is clearly a data entry error
})
clean_data = messy_data["city"].str.strip().str.lower()
print(clean_data)
```

```python
#5
print("Identifying outlier by quartile: ")
print(messy_data["score"].quantile(0.25))
print(messy_data["score"].quantile(0.75))
#The data below 85 and above 92 is most likely to be outlier so 1000 is almost certainly an error
```
