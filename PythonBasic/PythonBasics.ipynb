```python
"""
Build a persistent student database:

1. Write save_students(students, filename) that saves the students list 
   (list of dicts, like your Day 4 data) to a JSON file. Handle any 
   errors that could occur while writing (e.g., invalid path) — catch 
   them and print a clear message instead of crashing.
2. Write load_students(filename) that loads the students list back from 
   the JSON file. If the file doesn't exist yet (first run), return an 
   empty list [] instead of crashing — this is a common real pattern: 
   "if no saved data exists, start fresh."
3. Test the full cycle:
   - Call load_students("students.json") — should return [] since 
     the file doesn't exist yet
   - Save your Day 4 students list using save_students(...)
   - Call load_students(...) again — should now return your actual data
   - Print it to confirm it matches

   """
import json
students = [
    {"name": "Jack", "age": 19, "grades": [85, 90, 78]},
    {"name": "Priya", "age": 20, "grades": [92, 88, 95]},
    {"name": "John", "age": 24, "grades": [100, 99, 98]},
    {"name": "Alex", "age": 40, "grades": [70, 80, 90]}
]

def save_students(students, filename):
    try:
        with open(filename, "w") as f:
            json.dump(students, f)
    except OSError:
        print("Could not save file")

def load_students(filename):
    try:
        with open(filename, "r") as f:
            loaded_data = json.load(f)
            return loaded_data
    except FileNotFoundError:
        return []

loaded_data = load_students("student.json")
print(loaded_data)
save_students(students, "student.json")
loaded_data = load_students("student.json")
print(loaded_data)

```
